```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'DejaVu Serif'],
    'font.size': 10,
    'axes.labelsize': 10,
    'axes.titlesize': 10,
    'legend.fontsize': 10,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'text.usetex': False,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.02,
})

C_MAIN = '#2563EB'
C_HI   = '#059669'
C_LO   = '#DC2626'
C_DOT  = '#D97706'

def plot_panel(ax, nhi, acc, gpu, upper, lower, default_idx,
               upper_label, lower_label, title):
    ax.plot(nhi, acc, 'o-', color=C_MAIN, markersize=4, linewidth=1.5,
            label='DynaExq', zorder=5)
    ax.axhline(upper, color=C_HI, linestyle='--', linewidth=1.0,
               label=upper_label, zorder=3)
    ax.axhline(lower, color=C_LO, linestyle='--', linewidth=1.0,
               label=lower_label, zorder=3)
    # ax.plot(nhi[default_idx], acc[default_idx], '*', color=C_DOT,
    #         markersize=12, zorder=6, markeredgecolor='k', markeredgewidth=0.4)
    # ax.annotate(f'default ({nhi[default_idx]}%)',
    #             xy=(nhi[default_idx], acc[default_idx]),
    #             xytext=(nhi[default_idx] + 3, acc[default_idx] - 0.35),
    #             fontsize=10, color=C_DOT,
    #             arrowprops=dict(arrowstyle='->', color=C_DOT, lw=0.8))
    ax.set_xlabel('High-precision expert ratio (%)')
    ax.set_ylabel('Average accuracy (%)')
    # ax.set_title(title)
    ax.set_xlim(-2, 33)
    ax.set_ylim(min(lower, acc.min()) - 0.5, max(upper, acc.max()) + 0.5)
    ax.legend(loc='lower right', framealpha=0.9, edgecolor='#ccc')
    ax_top = ax.twiny()
    ax_top.set_xlim(ax.get_xlim())
    tick_pos = nhi[::2]
    tick_labels = [f'{gpu[i]:.0f}' for i in range(0, len(nhi), 2)]
    ax_top.set_xticks(tick_pos)
    ax_top.set_xticklabels(tick_labels)
    ax_top.set_xlabel('Device memory (GB)', fontsize=10, labelpad=4)
    ax_top.tick_params(labelsize=7.5)

panels = [
    dict(
        nhi=np.array([0, 5, 10, 15, 20, 25, 30]),
        acc=np.array([65.53, 66.38, 66.89, 67.12, 67.22, 67.28, 67.32]),
        gpu=np.array([33.8, 35.9, 37.9, 40.0, 42.1, 44.1, 46.2]),
        upper=67.73, lower=65.53, default_idx=4,
        upper_label='FP16 (all hi-prec)', lower_label='INT4 (all lo-prec)',
        title='(a) Qwen3-30B (FP16/INT4)',
        out='figures/budget_sensitivity_qwen30b.pdf',
    ),
    dict(
        nhi=np.array([0, 5, 10, 15, 20, 25, 30]),
        acc=np.array([73.58, 75.34, 76.37, 76.90, 77.15, 77.30, 77.40]),
        gpu=np.array([39.2, 40.1, 41.1, 42.0, 43.0, 43.9, 44.9]),
        upper=78.12, lower=73.58, default_idx=4,
        upper_label='INT4 (all hi-prec)', lower_label='INT2 (all lo-prec)',
        title='(b) Qwen3-80B (INT4/INT2)',
        out='figures/budget_sensitivity_qwen80b.pdf',
    ),
]

for p in panels:
    fig, ax = plt.subplots(1, 1, figsize=(6.4*0.5, 4.8*0.5))
    plot_panel(ax, p['nhi'], p['acc'], p['gpu'],
              p['upper'], p['lower'], p['default_idx'],
              p['upper_label'], p['lower_label'], p['title'])
    fig.tight_layout()
    fig.savefig(p['out'])
    print(f'Saved: {p["out"]}')
    plt.close(fig)

print('Done.')
```
